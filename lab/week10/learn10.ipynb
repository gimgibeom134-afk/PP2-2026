# 위젯 종류

기본 위젯(Widgets) 종류
* Label (라벨): 텍스트나 이미지를 화면에 표시하는 읽기 전용 위젯입니다.
* Button (버튼): 사용자가 클릭했을 때 특정 함수나 명령을 실행하는 버튼입니다.
* Entry (엔트리): 한 줄의 텍스트를 입력받을 수 있는 입력창입니다.
* Text (텍스트): 여러 줄의 텍스트를 입력하거나 출력할 수 있는 편집기 형태의 위젯입니다.
* Checkbutton (체크버튼): 여러 개 중 0개 이상을 중복해서 선택할 수 있는 체크박스입니다.
* Radiobutton (라디오버튼): 여러 개 중 단 하나의 옵션만 선택할 수 있는 라디오 버튼입니다.
* Listbox (리스트박스): 여러 줄의 항목 목록을 보여주고 선택하게 하는 목록 상자입니다.
* Combobox (콤보박스): ttk 모듈에 있는 항목 목록 드롭다운 선택창입니다.
* Scale (스케일): 슬라이더를 마우스로 끌어 연속된 숫자 값을 선택하는 위젯입니다.
* Scrollbar (스크롤바): 텍스트나 리스트박스, 캔버스 등에 연결해 화면을 스크롤하는 위젯입니다.
* Canvas (캔버스): 선, 도형, 이미지 등을 자유롭게 그리고 배치할 수 있는 도화지 영역입니다.
* Frame (프레임): 다른 위젯들을 담는 컨테이너(구조물) 역할의 영역입니다.
* Toplevel (탑레벨): 메인 창 외에 새로 생성되는 독립된 보조 창(팝업창)입니다.
* Menu / Menubutton (메뉴): 상단 메뉴바나 드롭다운 메뉴를 만드는 위젯입니다. [1, 2]

# 컨테이너 종류

* Fram프레임e () tk.Frame: 가장 기본적이고 널리 쓰이는 투명한 컨테이너입니다
* LabelFrame (라벨 프레임) tk.LabelFrame: Frame에 '테두리 선'과 '제목(라벨)'이 추가된 형태입니다
* Toplevel (탑레벨) tk.Toplevel: 메인 윈도우(root)와 별개로 완전히 독립된 새로운 보조 창(팝업창)을 띄우는 컨테이너입니다.
* PanedWindow (페인드 윈도우) tk.PanedWindow: 내부에 배치된 위젯들 사이에 마우스로 드래그하여 크기를 조절할 수 있는 구분선(경계선)을 제공
* Notebook (노트북 / 탭) ttk.Notebook: 웹 브라우저의 탭이나 책의 색인처럼 여러 개의 페이지(Tab)를 전환하며 볼 수 있게 해주는 컨테이너
입니다.(이 위젯은 일반 tk가 아닌 tkinter.ttk 모듈에 포함되어 있습니다.)
* Canvas (캔버스) tk.Canvas: 선, 원, 도형을 그리거나 이미지를 배치하는 도화지 위젯이지만, 일반 위젯을 내부에 얹을 수 있는 컨
테이너 역할도 겸합니다

# 숫자 추측 게임

```python
from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    guess = int(guessField.get())

    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음!"
    else:
        msg = "정답!"

    resultLabel["text"] = msg
    guessField.delete(0, 5)

def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 하세요!"

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")

window.geometry("500x80")
titleLabel = Label(window, text="숫자 게임에 오신 것을 환영합니다!",
bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side="left")
tryButton = Button(window, text="시도", fg="green", bg="white",
command=guessing )
tryButton.pack(side="left")

resetButton = Button(window, text="초기화", fg="red", bg="white",
command=reset)
resetButton.pack(side="left")
resultLabel = Label(window, text="1부터 100사이의 숫자를 입력하시오.",
bg="white")
resultLabel.pack(side="left")

window.mainloop()
```

# 게산기 프로그램

```python
from tkinter import *

window = Tk()
window.title("My Calculator")
display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
'7', '8', '9', '/', 'C',
'4', '5', '6', '*', ' ',
'1', '2', '3', '-', ' ',
'0', '.', '=', '+', ' ' ]

row_index = 1
col_index = 0

for button_text in button_list:
    Button(window, text=button_text, width=5).grid(row=row_index, column=col_index)
    col_index += 1
    if col_index > 4:
        row_index += 1
        col_index = 0
        
```
