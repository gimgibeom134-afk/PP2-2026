```python
import copy

temps = [[1,2], 3, 4, 5]
svalues = copy.copy(temps) #list(temps)와 동일
temps[0][0] = 100
temps
svalues

dvalues = copy.deepcopy(temps)

temps[0][0] = 7
temps
dvalues

```
